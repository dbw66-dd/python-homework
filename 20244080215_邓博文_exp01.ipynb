```python
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError("分数必须在0~100之间")
    if 90 <= score <= 100:
        return "优秀"
    elif 80 <= score < 90:
        return "良好"
    elif 60 <= score < 80:
        return "及格"
    else:
        return "不及格"

print(score_level(95))
print(score_level(85))
print(score_level(70))
print(score_level(50))

sum_for = 0
for i in range(1, 101):
    sum_for += i
print("for循环 1~100和：", sum_for)

sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1
print("while循环 1~100和：", sum_while)

def fact_iter(n):
    res = 1
    for k in range(2, n+1):
        res *= k
    return res

def fact_recur(n):
    if n == 0:
        return 1
    return n * fact_recur(n - 1)

for n in [0,1,5,10]:
    print(f"n={n},迭代阶乘={fact_iter(n)},递归阶乘={fact_recur(n)}")

def summary(values):
    min_v = min(values)
    max_v = max(values)
    avg_v = sum(values) / len(values)
    return (min_v, max_v, avg_v)

lst = [12,45,7,23,33]
min_num, max_num, avg_num = summary(lst)
print("最小值：", min_num, "最大值：", max_num, "平均值：", avg_num)

def sum_all(*args):
    return sum(args)

print("无参数：", sum_all())
print("单个参数：", sum_all(10))
print("多个参数：", sum_all(1,2,3,4,5))
```

```python
import os
import pickle

print(os.getcwd())
data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True)

content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
txt_path = os.path.join(data_dir, "students.txt")
with open(txt_path, "w", encoding="utf-8") as f:
    f.write(content)

with open(txt_path, "r", encoding="utf-8") as f:
    lines = f.readlines()[1:]

def parse_student(line):
    line = line.strip()
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError("字段数量错误")
    name, score_str = parts
    if len(name.strip()) == 0:
        raise ValueError("姓名为空")
    score = int(score_str)
    if not (0 <= score <= 100):
        raise ValueError("成绩超出0~100范围")
    return {"name":name, "score":score}

valid_list = []
error_log = []
for idx, line in enumerate(lines, start=2):
    try:
        rec = parse_student(line)
        valid_list.append(rec)
    except Exception as e:
        error_log.append((idx, str(e)))

valid_count = len(valid_list)
invalid_count = len(error_log)
scores = [d["score"] for d in valid_list]
avg_score = sum(scores)/len(scores)
pass_count = len([s for s in scores if s >=60])
pass_rate = pass_count / valid_count

report_path = os.path.join(data_dir, "report.txt")
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数：{valid_count}\n")
    f.write(f"无效记录数：{invalid_count}\n")
    f.write(f"平均分：{avg_score:.2f}\n")
    f.write(f"及格率：{pass_rate:.2%}\n")

with open(report_path, "a", encoding="utf-8") as f:
    f.write("追加内容演示：a模式不会覆盖原有文本\n")

pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_list, f)
with open(pkl_path, "rb") as f:
    loaded_data = pickle.load(f)
print(loaded_data == valid_list)

module_code = '''
def parse_student(line):
    line = line.strip()
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError("字段数量错误")
    name, score_str = parts
    if len(name.strip()) == 0:
        raise ValueError("姓名为空")
    score = int(score_str)
    if not (0 <= score <= 100):
        raise ValueError("成绩超出0~100范围")
    return {"name":name, "score":score}
'''
with open("student_utils.py","w",encoding="utf-8") as f:
    f.write(module_code)

from student_utils import parse_student
print(parse_student("Tom,78"))
try:
    parse_student("Tom,108")
except ValueError as e:
    print(e)

try:
    with open("notexist.txt","r") as f:
        pass
except FileNotFoundError:
    print("捕获异常：文件不存在")
```

```python
import copy

class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"


valid_list = [
    {"name": "Alice", "score": 85},
    {"name": "Bob", "score": 58},
    {"name": "Carol", "score": 92},
    {"name": "David", "score": 76},
    {"name": "Eva", "score": 60},
    {"name": "Frank", "score": 89}
]

stu_objs = []
for d in valid_list:
    s = Student(d["name"], d["score"])
    stu_objs.append(s)
    print(s, "及格" if s.is_passed() else "不及格")


class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"GraduateStudent(name={self.name}, score={self.score}, research_topic={self.research_topic})"

gstu = GraduateStudent("Li", 88, "机器学习")
print(gstu)


origin = {"name": "Alice", "scores": [85, 90]}

ref_obj = origin
ref_obj["scores"][0] = 100
print("直接赋值，原对象：", origin)

origin = {"name": "Alice", "scores": [85, 90]}
shallow = copy.copy(origin)
shallow["scores"][0] = 100
print("浅copy，原对象：", origin)

origin = {"name": "Alice", "scores": [85, 90]}
deep = copy.deepcopy(origin)
deep["scores"][0] = 100
print("深copy，原对象：", origin)
```
