```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

print(gcd(2018, 2019))
print(gcd(1200, 300))
print(gcd(5040, 60))
print(gcd(-1200, 300))
print(gcd(-1200, -300))

```

```python
import math

def euler_limit(n):
    return (1 + 1 / n) ** n



def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    fact = 1
    for k in range(n):
        if k == 0:
            fact = 1
        else:
            fact = fact * k
        total += 1 / fact
    return total



def euler_approx(epsilon):
    e_true = math.exp(1)
    total = 0.0
    fact = 1
    k = 0
    while abs(total - e_true) >= epsilon:
        if k == 0:
            fact = 1
        else:
            fact = fact * k
        total += 1 / fact
        k += 1
    return total


def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(val)



print(euler_limit(10000))
print(euler_infinite_sum(0))
print(euler_infinite_sum(1))
print(euler_infinite_sum(4))
print(euler_approx(0.001))
print_euler_sum_table(5)

def is_power_of_2(n):
    while n % 2 == 0:
        n = n // 2
    return n == 1

def is_power(b, n):
    if n == 1:
        return True
    while n % b == 0:
        n = n // b
    return n == 1

print(is_power_of_2(8))
print(is_power_of_2(10))
print(is_power(3,81))
print(is_power(2,12))
```
