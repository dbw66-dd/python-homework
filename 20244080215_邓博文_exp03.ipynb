```python
import re

# ========== 任务1 ==========
s1 = "cat on mat; later: cats and dogs"
res_d = re.findall(r"\d+", s1)
res_s = re.findall(r"\s+", s1)
res_cat = re.findall(r"\bcat\b", s1)
print("=====任务1=====")
print("\\d+ 匹配数字：", res_d)
print("\\s+ 匹配空白：", res_s)
print(r"\bcat\b 匹配独立单词cat：", res_cat)

# ========== 任务2 ==========
s2 = "cat dog bird CAT"
res_catdog = re.findall(r"cat|dog", s2)
res1 = re.findall(r"a|aa|aaa", "aaaa")
res2 = re.findall(r"aaa|aa|a", "aaaa")
print("\n=====任务2=====")
print("cat|dog：", res_catdog)
print(r"a|aa|aaa 匹配 aaaa：", res1)
print(r"aaa|aa|a 匹配 aaaa：", res2)

# ========== 任务3 ==========
s3 = "ct cat caat caaat"
p1 = re.findall(r"ca*t", s3)
p2 = re.findall(r"ca+t", s3)
p3 = re.findall(r"ca?t", s3)
p4 = re.findall(r"ca{2}t", s3)
p5 = re.findall(r"ca{1,2}t", s3)
print("\n=====任务3=====")
print("ca*t ：", p1)
print("ca+t ：", p2)
print("ca?t ：", p3)
print("ca{2}t：", p4)
print("ca{1,2}t：", p5)
```

```python
import re

# ========== 任务1：捕获组提取邮箱用户名、域名 ==========
s1 = "联系：alice@example.com；备选：bob@test.org"
pattern1 = r"(\w+)@(\w+\.\w+)"
res1 = re.findall(pattern1, s1)
print("=====任务1 邮箱（用户名，域名）=====")
print(res1)

# ========== 任务2：后向引用找连续重复单词 ==========
s2 = "cat cat, dog cat, bird bird"
pattern2 = r"\b(\w+)\s+\1\b"
res2 = re.findall(pattern2, s2)
print("\n=====任务2 连续重复单词=====")
print(res2)

# ========== 任务3：日期转换 年-月-日 → 月-日-年 ==========
s3 = "2024-05-01\n2024/05/02\n2024.05.03"
pattern3 = r"(\d{4})([-/.])(\d{2})([-/.])(\d{2})"
res3 = re.sub(pattern3, r"\3-\5-\1", s3)
print("\n=====任务3 日期转换=====")
print("转换前：")
print(s3)
print("转换后：")
print(res3)

# ========== 任务4：电话号码统一格式 (123)456-7890 ==========
s4 = "联系我：123-456-7890 或 987.654.3210"
pattern4 = r"(\d{3})[-.](\d{3})[-.](\d{4})"
res4 = re.sub(pattern4, r"(\1)\2-\3", s4)
print("\n=====任务4 号码格式化=====")
print("转换前：", s4)
print("转换后：", res4)
```

```python
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

soup = BeautifulSoup(html_doc, "html.parser")
title_text = soup.title.get_text()
h1_tag = soup.h1
h1_name = h1_tag.name
h1_text = h1_tag.get_text()
print("=====任务1=====")
print("页面标题文本：", title_text)
print("h1标签名：", h1_name)
print("h1文本：", h1_text)

print("\n=====任务2=====")
p_list = soup.find_all("p", class_="course")
print("所有class='course'的段落：")
for p in p_list:
    print(p.get_text())

a_list = soup.find_all("a")
print("\n所有链接：")
for a in a_list:
    link_text = a.get_text()
    link_href = a["href"]
    print(f"链接文字：{link_text}, href：{link_href}")

print("\n=====任务3=====")
all_text = soup.get_text()
p_text = p_list[0].get_text()
a_text = a_list[0].get_text()
print("整个文档文本：")
print(all_text)
print("单个段落文本：", p_text)
print("单个链接文本：", a_text)
```
