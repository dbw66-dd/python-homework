```python
import numpy as np


arr_int = np.array([1, 2, 3])
arr_float = np.array([1.1, 2.2, 3.3])
arr_bool = np.array([True, False, True])

print(f"整数数组：dtype={arr_int.dtype}, shape={arr_int.shape}, ndim={arr_int.ndim}")
print(f"浮点数组：dtype={arr_float.dtype}, shape={arr_float.shape}, ndim={arr_float.ndim}")
print(f"布尔数组：dtype={arr_bool.dtype}, shape={arr_bool.shape}, ndim={arr_bool.ndim}")
print("差别：NumPy ndarray要求全部元素是**相同数据类型**，存储紧凑、运算快；Python列表可以存放不同类型的元素。\n")


a = np.arange(12)
print("原始数组：", a)
a_34 = a.reshape(3, 4)
print("reshape(3,4)：\n", a_34)
a_232 = a.reshape(2, 3, 2)
print("reshape(2,3,2)：\n", a_232)
print("条件：新形状各维度数值相乘，必须等于数组总元素个数（12）\n")


arr3 = np.arange(12).reshape(3, 4)
print("基础数组：\n", arr3)
row2 = arr3[1, :]
print("取第2行：", row2, "  shape=", row2.shape)
col1_3 = arr3[:, 0:3]
print("取第1~3列：\n", col1_3, "  shape=", col1_3.shape)
greater6 = arr3[arr3 > 6]
print("大于6的元素：", greater6, "  shape=", greater6.shape, "\n")


arr4 = np.arange(10)
res_or = arr4[(arr4 < 2) | (arr4 > 7)]
res_and = arr4[np.logical_and(arr4 > 3, arr4 < 7)]
print("小于2或大于7：", res_or)
print("大于3且小于7：", res_and, "\n")


m1 = np.array([[1, 2], [3, 4]])
m2 = np.array([[5, 6], [7, 8]])
print("m1:\n", m1)
print("m2:\n", m2)
elem1 = m1 * m2
elem2 = np.multiply(m1, m2)
mat1 = m1 @ m2
mat2 = np.matmul(m1, m2)
print("* 逐元素乘法：\n", elem1)
print("np.multiply 逐元素乘法：\n", elem2)
print("@ 矩阵乘法：\n", mat1)
print("np.matmul 矩阵乘法：\n", mat2)
print("区别：* 和 np.multiply() 是**对应位置元素相乘**；@ 和 np.matmul() 是线性代数标准矩阵乘法。")
```

```python
import numpy as np

# ------------------------
np.random.seed(42)
scores = np.random.normal(loc=70, scale=10, size=1000)
scores = np.clip(scores, 0, 100)
n = len(scores)
min_s = scores.min()
max_s = scores.max()
mean_s = scores.mean()
std_s = scores.std()
ptp_s = np.ptp(scores)
print(f"样本数:{n}, 最小值:{min_s:.2f}, 最大值:{max_s:.2f}")
print(f"均值:{mean_s:.2f}, 标准差:{std_s:.2f}, 极差:{ptp_s:.2f}")

# ------------------------
arr = np.arange(1, 11)
sample_uniform = np.random.choice(arr, size=5, replace=False)
p = arr / arr.sum()
sample_weighted = np.random.choice(arr, size=5, replace=False, p=p)
print(f"原数组: {arr}")
print(f"均匀不放回抽样: {sample_uniform}")
print(f"加权抽样(权重等于数值): {sample_weighted}")

# ------------------------
x = np.array([1, 2, 3, 4, 5])
x_copy = x.copy()
shuffled_new = np.random.permutation(x_copy)
print(f"permutation前原数组: {x_copy}")
print(f"permutation得到新数组: {shuffled_new}")
print(f"permutation后原数组: {x_copy}")

x_inplace = np.array([1, 2, 3, 4, 5])
print(f"\nshuffle前原数组: {x_inplace}")
np.random.shuffle(x_inplace)
print(f"shuffle后原数组: {x_inplace}")

# ------------------------
scores_nan = scores.copy()
scores_nan[[9, 19, 29]] = np.nan
mean_normal = np.mean(scores_nan)
mean_nan = np.nanmean(scores_nan)
nan_count = np.isnan(scores_nan).sum()
print(f"含nan数组普通mean: {mean_normal}")
print(f"忽略nan的nanmean: {mean_nan:.2f}")
print(f"缺失值个数: {nan_count}")
print("说明：普通np.mean遇到nan会直接返回nan；np.nanmean会跳过nan计算均值")

# ------------------------
idx_arr = np.arange(len(scores_nan))
table = np.rec.fromarrays([idx_arr, scores_nan], names=["index", "score"])
print("内存表格前10行：")
print(table[:10])
```

```python
import numpy as np
from scipy.linalg import solve
from scipy.optimize import root, minimize
from scipy.stats import ttest_ind
import matplotlib.pyplot as plt
plt.rcParams['figure.dpi'] = 100
# 新增解决中文乱码
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

# ------------------------
# 1. 解线性方程组 3x + y = 9, x + 2y = 8
A = np.array([[3, 1],
              [1, 2]])
b = np.array([9, 8])
sol = solve(A, b)
x, y = sol
print(f"方程组解：x={x:.2f}, y={y:.2f}")
print(f"代入验证 3x+y = {3*x + y:.2f}, x+2y = {x + 2*y:.2f}")

# ------------------------
# 2. root 求解 x² - 2 = 0
def f(x):
    return x**2 - 2
res_pos = root(f, x0=1.5)
res_neg = root(f, x0=-1.5)
print("正根求解成功：", res_pos.success, " 结果：", res_pos.x, " f值：", f(res_pos.x))
print("负根求解成功：", res_neg.success, " 结果：", res_neg.x, " f值：", f(res_neg.x))

# ------------------------
# 3. minimize 求 f(x)=x²+x+2 最小值
def func(x):
    return x**2 + x + 2
res_min = minimize(func, x0=0)
print("极小点：", res_min.x, " 极小值：", res_min.fun)
print("解析解：x=-0.5，f=1.75")

# ------------------------
# 4. 独立样本t检验
np.random.seed(42)
s1 = np.random.normal(loc=0, scale=1, size=100)
s2 = np.random.normal(loc=1, scale=1, size=100)
t_stat, p_val = ttest_ind(s1, s2)
print(f"t统计量={t_stat:.4f}, p值={p_val:.4f}")
if p_val < 0.05:
    print("p<0.05，拒绝原假设，两组均值有显著差异")
else:
    print("p>=0.05，不能拒绝原假设，两组均值无显著差异")

# ------------------------
# 5. 绘图
np.random.seed(42)
scores = np.random.normal(loc=70, scale=10, size=1000)
scores = np.clip(scores, 0, 100)
scores_nan = scores.copy()
scores_nan[[9,19,29]] = np.nan
valid_scores = scores_nan[~np.isnan(scores_nan)]

# 直方图
plt.figure()
plt.hist(valid_scores, bins=20)
plt.title("成绩直方图")
plt.xlabel("成绩")
plt.ylabel("频数")
plt.show()

# 箱线图
plt.figure()
plt.boxplot(valid_scores)
plt.title("成绩箱线图")
plt.ylabel("成绩")
plt.show()

# 折线图 y=x, y=x², y=x³
plt.figure()
x_line = np.linspace(0,5,100)
plt.plot(x_line, x_line, label="$y=x$")
plt.plot(x_line, x_line**2, label="$y=x^2$")
plt.plot(x_line, x_line**3, label="$y=x^3$")
plt.title("函数折线图")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np

# ------------------------
records = {
    "Name": ["Alice", "Bob", "Carol", "David", "Eva", "Frank"],
    "Group": ["A", "A", "B", "B", "A", "B"],
    "Math": [88, 72, 91, 65, 79, 84],
    "English": [90, 68, 85, 73, 92, 77],
    "Python": [95, 80, 89, 70, 86, 82],
}
df = pd.DataFrame(records)
df = df.set_index("Name")
df["Total"] = df["Math"] + df["English"] + df["Python"]
df["Average"] = df["Total"] / 3
print("行索引：", df.index.tolist())
print("列名：", df.columns.tolist())
print("\n前3行：")
print(df.head(3))
print("\n基本统计信息：")
print(df.describe())

# ------------------------
print("\nCarol记录 loc：")
print(df.loc["Carol"])
print("\nCarol记录 iloc：")
print(df.iloc[2])
print("\n平均分不低于80的学生：")
print(df[df["Average"] >= 80])

# ------------------------
print("\n各科列均值 mean(axis=0)：")
print(df[["Math","English","Python"]].mean(axis=0))
print("\n每名学生行均值 mean(axis=1)：")
print(df[["Math","English","Python"]].mean(axis=1))

# ------------------------
df["MaxScore"] = df[["Math","English","Python"]].apply(max, axis=1)
print("\n每科 min,mean,max agg：")
print(df[["Math","English","Python"]].agg(["min","mean","max"]))
df["Math_Rank"] = df["Math"].rank(ascending=False)
print("\n数学成绩排名：")
print(df[["Math","Math_Rank"]])

# ------------------------
group_stat = df.groupby("Group").agg(人数=("Group","count"), 三科平均分=("Average","mean"))
print("\n分组统计：")
print(group_stat)
df["Group_Python_Mean"] = df.groupby("Group")["Python"].transform("mean")
print("\n增加组内Python平均分：")
print(df)

# ------------------------
group_filtered = df.groupby("Group").filter(lambda g: g["Python"].mean() >= 80)
print("\nPython组平均分>=80的组及学生：")
print(group_filtered)

# ------------------------
score_cols = ["Math","English","Python"]
cov_mat = df[score_cols].cov()
corr_mat = df[score_cols].corr()
print("\n协方差矩阵：")
print(cov_mat)
print("\n皮尔森相关系数矩阵：")
print(corr_mat)

print("\n解释示例（以Math和English为例）：")
print("协方差为正：说明数学成绩高的人，英语成绩倾向也偏高，二者同向变化；")
print("皮尔森相关系数大小：越接近1代表线性正相关越强，越接近0代表线性关联越弱。")
```
