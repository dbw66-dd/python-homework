```python
def is_palindrome(s):
    clean = s.replace(" ", "").lower()
    return clean == clean[::-1]


def is_abecedarian(s):
    clean = s.replace(" ", "").lower()
    for i in range(len(clean) - 1):
        if clean[i] > clean[i+1]:
            return False
    return True


def remove_vowels(s):
    vowels = {"a", "e", "i", "o", "u"}
    res = ""
    for ch in s:
        if ch.lower() not in vowels:
            res += ch
    return res


print(is_palindrome("T A C O cat"))
print(is_abecedarian("abcd"))
print(remove_vowels("audio"))
```

```python
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表")
    return lst[::-1]


def is_sorted(seq):
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False
    return True


print(list_reverse([1, 2, 3]))
print(is_sorted([1, 2, 3, 4]))
print(is_sorted([1, 3, 2]))
```

```python
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    if not isinstance(s, (int, float)):
        raise TypeError("标量必须是整数或浮点数")
    for elem in t:
        if not isinstance(elem, (int, float)):
            raise TypeError("元组内元素必须是整数或浮点数")
    return tuple(x * s for x in t)


def vec_inner_product(v1, v2):
    if not isinstance(v1, tuple) or not isinstance(v2, tuple):
        raise TypeError("输入必须是元组")
    if len(v1) != len(v2):
        raise ValueError("两个向量长度不一致")
    for elem in v1:
        if not isinstance(elem, (int, float)):
            raise TypeError("向量元素必须是整数或浮点数")
    for elem in v2:
        if not isinstance(elem, (int, float)):
            raise TypeError("向量元素必须是整数或浮点数")
    total = 0.0
    for a, b in zip(v1, v2):
        total += a * b
    return total


def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if len(mx) == 0:
        return True
    row_len = len(mx[0])
    for row in mx:
        if not isinstance(row, tuple):
            return False
        if len(row) != row_len:
            return False
        for val in row:
            if not isinstance(val, (int, float)):
                return False
    return True


def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise TypeError("矩阵格式不合法")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    for e in vec:
        if not isinstance(e, (int, float)):
            raise TypeError("向量元素必须是整数或浮点数")
    if len(mx) == 0:
        return ()
    cols = len(mx[0])
    if len(vec) != cols:
        raise ValueError("矩阵列数和向量长度不匹配")
    res = []
    for row in mx:
        res.append(vec_inner_product(row, vec))
    return tuple(res)


if __name__ == "__main__":
    print(vec_scalar_mult((1, 2, 3), 2))
    print(vec_inner_product((1,2,3), (4,5,6)))
    M = ((1,2), (3,4))
    print(check_valid_mx(M))
    print(mx_vec_mult(M, (1, 1)))
```

```python
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        if elements is None:
            elements = [0.0] * dim
        if len(elements) != dim:
            raise ValueError("维度和元素数量不一致")
        for e in elements:
            if not isinstance(e, (int, float)):
                raise TypeError("元素只能是整数或浮点数")
        self.dim = dim
        self.elements = tuple(elements)

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return False
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的向量")
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的向量")
        return self.elements < other.elements

    def __le__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的向量")
        return self.elements <= other.elements

    def __gt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的向量")
        return self.elements > other.elements

    def __ge__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的向量")
        return self.elements >= other.elements

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("点积的参数必须是Vector")
        if self.dim != other.dim:
            raise ValueError("两个向量维度不一致，无法计算内积")
        return sum(a * b for a, b in zip(self.elements, other.elements))

    def __mul__(self, other):
        if isinstance(other, (int, float)):
            new_elems = [x * other for x in self.elements]
            return Vector(self.dim, new_elems)
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("向量维度不一致，不能逐元素相乘")
            new_elems = [a * b for a, b in zip(self.elements, other.elements)]
            return Vector(self.dim, new_elems)
        else:
            raise TypeError("不支持该类型乘法")

    def __rmul__(self, other):
        return self * other

    def __repr__(self):
        return f"Vector({self.dim}, {list(self.elements)})"


if __name__ == "__main__":

    print("第2题答案：")
    print("维度和元素应当是实例属性。")
    print("理由：每个Vector实例代表一个独立向量，不同向量可以拥有不同的维度与不同的元素，")
    print("维度、元素是属于单个向量个体的数据，而不是所有Vector对象共享的统一属性，因此使用实例属性。")


    v1 = Vector(2, [2, 4])
    v2 = Vector(2, [2, 5])
    print("\n测试：")
    print("v1 < v2:", v1 < v2)
    print("v1.dot(v2):", v1.dot(v2))
    print("v1 * 2:", v1 * 2)
    print("2 * v1:", 2 * v1)
    print("v1 * v2:", v1 * v2)
```
